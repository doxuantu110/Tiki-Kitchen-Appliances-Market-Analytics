### 4. Primary key & Duplicate Check

#### 4.1. Product ID

```python
def check_key(df, column):
    return {
        "total_rows": len(df),
        "missing": df[column].isna().sum(),
        "unique": df[column].nunique(),
        "duplicates": df[column].duplicated().sum()
    }
```

```python
product_ids_quality = check_key(products,"id")

product_ids_quality
```

#### 4.2. Product IDs vs Product

```python
source_ids = set(
    product_ids["product_id"]
    .dropna()
    .astype(str)
)

product_data_ids = set(
    products["id"]
    .dropna()
    .astype(str)
)

print("IDs collected:", len(source_ids))
print("IDs in product data:", len(product_data_ids))

print("Missing product data:",
      len(source_ids - product_data_ids))

print("Unexpected product IDs:",
      len(product_data_ids - source_ids))
```

#### 4.3. Review ID

```python
review_id_quality = check_key(reviews, "comment_id")

review_id_quality
```

### 5. Missing value Analysis

```python
def missing_summary(df):
    result = pd.DataFrame({
        "missing_count": df.isna().sum(),
        "missing_pct": df.isna().mean() * 100
    })

    return (
        result
        .sort_values("missing_pct", ascending=False)
    )

product_missing = missing_summary(products)

display(product_missing)
```

```python
display(product_missing[product_missing["missing_count"]>0])
```

#### Missing status

```python
def missing_status(pct):
    if pct == 0:
        return "OK"
    elif pct < 1:
        return "Low"
    elif pct < 5:
        return "Moderate"
    else:
        return "High"


product_missing["status"] = (
    product_missing["missing_pct"]
    .apply(missing_status)
)

display(product_missing)
```

#### 7. Cross-Dataset Consistency

```python
products_with_reviews = set(
    products.loc[
        products["review_count"] > 0,
        "id"
    ].astype(str)
)

review_product_ids = set(
    reviews["product_id"]
    .dropna()
    .astype(str)
)

print(
    "Products with review_count > 0:",
    len(products_with_reviews)
)

print(
    "Products appearing in review dataset:",
    len(review_product_ids)
)
```

```python
missing_review_products = (
    products_with_reviews -
    review_product_ids
)

unexpected_review_products = (
    review_product_ids -
    products_with_reviews
)

print(
    "Products expected to have reviews but missing:",
    len(missing_review_products)
)

print(
    "Products with reviews but review_count = 0:",
    len(unexpected_review_products)
)
```

```python
crawled_review_counts = (
    reviews
    .groupby("product_id")
    .size()
    .reset_index(name="crawled_review_count")
)

review_quality = products.merge(
    crawled_review_counts,
    left_on="id",
    right_on="product_id",
    how="left"
)

review_quality["crawled_review_count"] = (
    review_quality["crawled_review_count"]
    .fillna(0)
    .astype(int)
)
```

```python
review_quality[
    [
        "product_id",
        "review_count",
        "crawled_review_count"
    ]
].head(20)
```

```python
review_quality["review_crawl_coverage"] = np.where(
    review_quality["review_count"] > 0,
    review_quality["crawled_review_count"] /
    review_quality["review_count"],
    np.nan
)
```

```python
invalid_review_rating = reviews[
    (reviews["rating"] < 1) |
    (reviews["rating"] > 5)
]

print(
    "Invalid review ratings:",
    len(invalid_review_rating)
)
```

#### 8. Data Quality Summary

```python
quality_summary = pd.DataFrame([
    {
        "category": "Completeness",
        "check": "Product ID missing",
        "invalid_records": product_ids["product_id"].isna().sum()
    },
    {
        "category": "Uniqueness",
        "check": "Duplicate product IDs",
        "invalid_records": products["id"].duplicated().sum()
    },
    {
        "category": "Consistency",
        "check": "Product IDs missing from product data",
        "invalid_records": len(source_ids - product_data_ids)
    },
    {
        "category": "Consistency",
        "check": "Unexpected product IDs",
        "invalid_records": len(product_data_ids - source_ids)
    },
    {
        "category": "Validity",
        "check": "Invalid prices",
        "invalid_records": len(
            products[products["price"] <= 0]
        )
    },
    {
        "category": "Validity",
        "check": "Invalid ratings",
        "invalid_records": len(
            products[
                (products["rating_average"] < 0) |
                (products["rating_average"] > 5)
            ]
        )
    },
    {
        "category": "Consistency",
        "check": "Products with reviews but missing review data",
        "invalid_records": len(missing_review_products)
    }
])

display(quality_summary)
```
